Para fazer a leitura dos dados do arquivo csv precisamos importar a biblioteca csv, nele usaremos o metodo csv.reader para a leitura dos dados

```python
# Importando a biblioteca csv
import csv
```

```python
# Carregando os dados do csv e adicionando em um variavel
dados_csv = [] # Lista vazia
with open(path_csv, 'r') as file:
    dados = csv.reader(file, delimiter=',')
    # Ao visualizar os dados vemos que retorna um object, isso porque ele precisa ser 
    # lido dado a dado, sendo iteravel, podemos usar um laço for para visualizar os dados
    # Usando o laço for
    for dado in dados:
        # Os dados irão vir em grande volume em listas uma a uma, vamos usar o append para
        # adicionar os dados em uma lista vazia, que esta fora da função
        dados_csv.append(dado)
```

```python
# Visualizando os dados da primeira lista
dados_csv[0]
```

Agora temos uma lista de listas, sendo necessario transformar em um dicionario, assim com o dados em json, na biblioteca csv temos o metodo dictreader para retornar um dicionario

```python
# Da mesma forma que usamos para criar a lista de listas, criaremos uma lista de dict
dados_csv = [] # Lista vazia
with open(path_csv, 'r') as file:
    dados = csv.DictReader(file, delimiter=',')
    # Ao visualizar os dados vemos que retorna um object, isso porque ele precisa ser 
    # lido dado a dado, sendo iteravel, podemos usar um laço for para visualizar os dados
    # Usando o laço for
    for dado in dados:
        # Os dados irão vir em grande volume em listas uma a uma, vamos usar o append para
        # adicionar os dados em uma lista vazia, que esta fora da função
        dados_csv.append(dado)
```

Para resolver o erro na hora de salvar o arquivo, usando o dictwrite, vamos tratar os dados usando um metodo get

```python
# primeiro vamos trazer a coluna que tenha todos os nomes
colunas_combinadas = list(dados_agrupados[-1].keys())
colunas_combinadas
```

```python
# Criamos uma lista para receber os nomes das colunas 
dados_combinados_tabela = [colunas_combinadas]

# No for será o tratamento na qual ele passa por cada linha, e caso não tenha o valor da 
# data de venda, o get insere o valor de indisponivel, e depois adiciona a lista vazia
for i in dados_agrupados:
    linha = []
    for coluna in colunas_combinadas:
        linha.append(i.get(coluna, 'Indisponivel'))
    dados_combinados_tabela.append(linha)
```

```python
dados_combinados_tabela[0]
```

```python
# Agora vamos subescrever o arquivo csv
path_dados_agrupados = '../data_processed/dados_combinados.csv'

with open(path_dados_agrupados, 'w') as file:
    writer = csv.writer(file)
    writer.writerows(dados_combinados_tabela)
```
